### **Introduction: Applying Activation Functions to Financial Data in Google Colab**  

Activation functions play a crucial role in deep learning models by introducing non-linearity, allowing neural networks to learn complex patterns. In this Google Colab notebook, we apply two commonly used activation functions—**Sigmoid** and **ReLU (Rectified Linear Unit)**—to real-world financial data using **Yahoo Finance**.  

The dataset consists of historical stock prices for **Apple Inc. (AAPL)**, retrieved using the `yfinance` library. To ensure compatibility with activation functions, we **normalize** the closing price data to a range between **-10 and 10**, mimicking the type of inputs typically seen in deep learning models.  

We then apply the **Sigmoid** function, which maps values between **0 and 1**, and the **ReLU** function, which retains positive values while setting negative inputs to zero. By visualizing the transformed stock price data, we gain insights into how activation functions process financial time series and their potential applications in **quantitative finance, algorithmic trading, and deep learning-based stock prediction**.  

This notebook provides an intuitive way to understand activation functions beyond theoretical definitions, linking them to real-world stock market data for a more practical perspective.

```python
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf
from sklearn.preprocessing import MinMaxScaler
```

```python

# Fetch stock data (AAPL for the last 6 months)
stock = yf.download("AAPL", period="6mo", interval="1d")
closing_prices = stock["Close"].values.reshape(-1, 1)

# Normalize stock prices to fit between -10 and 10
scaler = MinMaxScaler(feature_range=(-10, 10))
normalized_prices = scaler.fit_transform(closing_prices).flatten()

# Define activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

# Compute function values
y_sigmoid = sigmoid(normalized_prices)
y_relu = relu(normalized_prices)
```

```python
# Plot Sigmoid
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.plot(y_sigmoid, label="Sigmoid", color="blue")
plt.title("Sigmoid on Stock Prices")
plt.xlabel("Days")
plt.ylabel("Activation Output")
plt.grid()
plt.legend()

# Plot ReLU
plt.subplot(1, 2, 2)
plt.plot(y_relu, label="ReLU", color="red")
plt.title("ReLU on Stock Prices")
plt.xlabel("Days")
plt.ylabel("Activation Output")
plt.grid()
plt.legend()

# Show plots
plt.tight_layout()
plt.show()
```
